Vocabulary and text exploration use training data only. Test transcripts remain held out.
Source text and reviewed Mendeley abbreviations are preserved in the canonical manifests.

```python
import pandas as pd
import numpy as np
```

```python
from pathlib import Path

root = next((parent / "data" for parent in (Path.cwd(), *Path.cwd().parents)
             if (parent / "data/karakalpak-combined/train.csv").is_file()), None)
if root is None:
    raise FileNotFoundError("karakalpak-combined/train.csv")
data = pd.read_csv(root / "karakalpak-combined/train.csv")
```

```python
data.head()
```

```python
unique = []

for i in range(len(data["text"])):
    for text in data["text"][i].split():
        for j in text:
            if j not in unique and j.isalpha():
                unique.append(j)
print(len(unique))
print(unique)
```

```python
import unicodedata

def clean_sentence(text):
    text = unicodedata.normalize("NFC", str(text))
    return " ".join("".join(c if c.isalnum() or c.isspace() else " " for c in text).split())

data["text"] = data["text"].apply(clean_sentence)
```

```python
data.head()
```

```python
def to_lowercase(text):
    return str(text).lower()

data["text"] = data["text"].apply(to_lowercase)
```
