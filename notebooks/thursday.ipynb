```python
import re
import unicodedata

ALPHABET = (
    "AÁBDEFGǴHXÍIJKQLMNŃOÓPRSTUÚVWYZC"
    "aábdefgǵhxıijkqlmnńoóprstuúvwyzc"
)

def clean_sentence(text):
    text = unicodedata.normalize("NFC", str(text))

    # Remove punctuation but keep Karakalpak letters, digits, and spaces
    text = re.sub(
        rf"[^{re.escape(ALPHABET)}0-9\s]",
        " ",
        text
    )

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


data["sentence_latin"] = data["sentence_latin"].apply(clean_sentence)
```

```python
def to_lowercase(text):
    return str(text).lower()

data["sentence_latin"] = data["sentence_latin"].apply(to_lowercase)
```
